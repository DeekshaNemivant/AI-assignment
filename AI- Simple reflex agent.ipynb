```python
pm25_breakpoints = [
    (0.0, 9.0, 0, 50),
    (9.1, 35.4, 51, 100),
    (35.5, 55.4, 101, 150),
    (55.5, 125.4, 151, 200),
    (125.5, 225.4, 201, 300),
    (225.5, 325.4, 301, 500)
]
```

```python
def calculate_aqi(C, breakpoints):

    for Clow, Chigh, Ilow, Ihigh in breakpoints:

        if Clow <= C <= Chigh:

            AQI = ((Ihigh - Ilow) / (Chigh - Clow)) * \
                  (C - Clow) + Ilow

            return AQI

    return np.nan
```

```python
def classify_aqi(aqi):
    if aqi <= 50:
        return "Good"
    elif aqi <= 100:
        return "Moderate"
    elif aqi <= 150:
        return "Poor"
    elif aqi <= 200:
        return "Severe"
    else:
        return "Very Severe"
```

```python
df['aqi_category'] = df['calculated_aqi'].apply(classify_aqi)
```
