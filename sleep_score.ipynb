```python
import pandas as pd

df = pd.read_csv("sleep_health_dataset.csv")

print(df.head())
```

```python
df["core_percentage"] = (
    100
    - df["rem_percentage"]
    - df["deep_sleep_percentage"]
)
```

```python
df["rem_score"] = (
    100
    - abs(df["rem_percentage"] - 22.5) / 22.5 * 100
)
df["rem_score"] = df["rem_score"].clip(lower=0)
df["rem_score"]
```

```python
df["deep_score"] = (
    100
    - abs(df["deep_sleep_percentage"] - 17.5) / 17.5 * 100
)
df["deep_score"] = (
    100
    - abs(df["deep_sleep_percentage"] - 17.5) / 17.5 * 100
)
df["deep_score"]
```

```python
df["core_score"] = (
    100
    - abs(df["core_percentage"] - 60) / 60 * 100
)
df["core_score"] = df["core_score"].clip(lower=0)
df["core_score"]
```

```python
df["awake_score"] = (
    100
    - df["wake_episodes_per_night"] * 15
)
df["awake_score"] = df["awake_score"].clip(lower=0)
df["awake_score"]
```

```python
df["sleep_score"] = (
    df["rem_score"] * 0.25
    + df["deep_score"] * 0.25
    + df["core_score"] * 0.25
    + df["awake_score"] * 0.25
)
df["sleep_score"] = df["sleep_score"].round(2)
```

```python
def sleep_category(score):

    if score >= 90:
        return "Excellent"

    elif score >= 80:
        return "Good"

    elif score >= 70:
        return "Fair"

    elif score >= 60:
        return "Poor"

    else:
        return "Very Poor"
```

```python
df["sleep_category"] = df["sleep_score"].apply(sleep_category)
```
